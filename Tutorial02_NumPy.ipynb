```python
import numpy as np
```

**TASK 1**

```python
a = np.array([[1, 2],
              [3, 4]])
print(a)
print("Shape:", a.shape)
```

```python
rand_arr = np.random.rand(3, 4)
print(rand_arr)
print("Shape:", rand_arr.shape)
```

**TASK 3**

```python
ones_arr = np.ones((2, 3))
zeros_arr = np.zeros((2, 3))
print("Ones:\n", ones_arr)
print("Zeros:\n", zeros_arr)
```

**TASK 4**

```python
arr = np.random.rand(6, 3)
print("Original shape:", arr.shape)
print(arr)

arr_T = arr.T
print("Transposed shape:", arr_T.shape)
print(arr_T)
```

**TASK 5**

```python
n = 4
p = 5
m = 6

A = np.random.rand(n, p)
B = np.random.rand(p, m)

C = np.dot(A, B)

print("A shape:", A.shape)
print("B shape:", B.shape)
print("Result:\n", C)
print("Result shape:", C.shape)
```
